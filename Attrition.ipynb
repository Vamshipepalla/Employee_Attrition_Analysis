```python
df['experience_level'] = pd.cut(df['experience'],bins = [0,3,10,26] , labels = ['Fresher','Senior','Expert'], include_lowest = True)

# Freshers
# 0 is Exclusive      -----> cab be included using include_lowest = True
# 3 is inclusive

# Senior
# 3 is Exclusive
# 10 is inclusive

# Expert
# 10 is Exclusive
# 26 is Inclusive
```

```python
def job_clean(row):
    j = row['job']

    if 'Director' in j:
        return 'Director'
    elif 'Exec' in j:
        return j.removeprefix("Exec").replace(",","").strip()
    elif 'Manager' in j:
        return 'Manager'
```

```python
hire_df = (df['hire_date'].dt.year).value_counts().sort_index(ascending = True).reset_index()

hire_df = hire_df.rename(columns = {'hire_date': 'hired_year','count' : 'hirings'})
```

```python
term_df = (df['term_date'].dt.year).value_counts().sort_index(ascending = True).reset_index()

term_df = term_df.rename(columns = {'term_date': 'term_year','count' : 'term_cnt'})
```
